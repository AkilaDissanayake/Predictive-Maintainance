## RUL Prediction

```python
from utils.Timestamp_handing import process_timestamps
#lets split the data into features and target variable
Target_columns=['RUL','TTF','Component_Health_Score','Maintenance_Type','Failure_Probability']
X = df.drop(columns=Target_columns)
#handle Timestamp column
X = process_timestamps(X, col_name='Timestamp')
Y= df[['RUL']] #[[]] TO MAKE Y A DATAFRAME INSTEAD OF SERIES



```

```python
#Lets train a model only on first 50000 rows to see if it can learn something
X_train = X.iloc[:50000]
Y_train = Y.iloc[:50000]

X_val = X.iloc[50000:62000]
Y_val = Y.iloc[50000:62000]

X_test = X.iloc[62000:74000]
Y_test = Y.iloc[62000:74000]

print("--- Training Data Types ---")
print(X_train.dtypes)


```

```python
from utils.Preproccessing import normalize_dataframe
#Normalize the features
# Normalize X_train and get the scaler
X_train_norm, X_train_scaler = normalize_dataframe(X_train)
Y_train_norm, Y_train_scaler = normalize_dataframe(Y_train)
# Normalize X_val and X_test using the same scaler
X_val_norm = X_train_scaler.transform(X_val)
Y_val_norm = Y_train_scaler.transform(Y_val)

#Normalize X_test and Y_test using the same scaler
X_test_norm = X_train_scaler.transform(X_test)
Y_test_norm = Y_train_scaler.transform(Y_test)
```

```python
# 1. Check Data Types (Should all be float32 or float64)
print("--- Data Types Post-Normalization ---")
print(f"X_train_norm type: {type(X_train_norm)}")
print(f"Y_train_norm type: {type(Y_train_norm)}")
print(f"X_val_norm type:   {type(X_val_norm)}")
print(f"Y_val_norm type:   {type(Y_val_norm)}")

# 2. Check Shapes (Crucial for the Input Layer)
print("\n--- Shape Audit ---")
print(f"X_train_norm shape: {X_train_norm.shape}") # Expect (10000, 27)
print(f"Y_train_norm shape (Raw): {Y_train_norm.shape}")
```

**Traansform convert df to ndarray use ravel**

```python
from utils.Models import build_mlp_model
mlp_model=build_mlp_model(input_shape=(X_train.shape[1],))
mlp_model.summary()
```

```python
from tensorflow.keras.optimizers import Adam
from utils.Train import train_rul_model
import numpy as np

opt = Adam(learning_rate=0.0001, clipvalue=1.0)

# 2. Call the training function using  NORMALIZED data

history, model_path = train_rul_model(
    model=mlp_model, 
    X_train=X_train_norm, 
    y_train=np.ravel(Y_train_norm), 
    X_val=X_val_norm, 
    y_val=np.ravel(Y_val_norm), 
    epochs=100, 
    batch_size=64,
    loss='huber',           # Robust against outliers and prevents exploding gradients
    optimizer=opt,          # Our custom stable optimizer
    model_name="mlp_rul_model"
)
```

```python
from utils.Plots import plot_history
plot_history(history, title="MLP RUL Model Training History")
```
